```python
import numpy as np
import matplotlib.pyplot as plt

# ─── Parameters ───────────────────────────────────────────────
g = 9.81
m = 0.145
dt = 0.01
t_max = 5.0

# ─── Force function ───────────────────────────────────────────
def compute_acceleration(x, y, vx, vy):
    """Return (ax, ay) given current state."""
    ax = 0.0
    ay = -g
    return ax, ay

# ─── Initial conditions ───────────────────────────────────────
x, y = 0.0, 0.0
vx, vy = 15.0, 25.0

# ─── Storage ──────────────────────────────────────────────────
trajectory = {'t': [0], 'x': [x], 'y': [y], 'vx': [vx], 'vy': [vy], 'KE': [0.5*m*(vx**2+vy**2)], 'PE': [m*g*y], 'E': [0.5*m*(vx**2+vy**2)+m*g*y] }

# ─── Simulation loop ──────────────────────────────────────────
t = 0.0
while t < t_max and y >= 0:
    ax, ay = compute_acceleration(x, y, vx, vy)
    vx_old, vy_old = vx, vy   # save old velocity
    vx += ax * dt
    vy += ay * dt
    x += vx * dt
    y += vy * dt
    t += dt
    KE = 0.5 * m * (vx**2 + vy**2)
    PE = m * g * y
    E  = KE + PE
    for key, val in zip(['t','x','y','vx','vy','KE','PE','E'], [t,x,y,vx,vy,KE,PE,E]):
        trajectory[key].append(val)

# ─── Visualization ────────────────────────────────────────────
plt.plot(trajectory['x'], trajectory['y'])
plt.xlabel('x (m)'); plt.ylabel('y (m)')
plt.title('Trajectory'); plt.grid(True, alpha=0.3)
plt.show()
```
