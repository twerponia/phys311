```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def simulate_freefall(h, dt, g=9.81):
    """Simulate 1D free fall using the Euler method.
    
    Parameters:
        h  : initial height (m)
        dt : time step (s)
        g  : gravitational acceleration (m/s²)
    
    Returns:
        t_arr, y_arr, vy_arr : lists of time, height, velocity
    """
    y, vy = h, 0.0
    t = 0.0
    
    t_arr, y_arr, vy_arr = [t], [y], [vy]
    
    while y >= 0:
        # Compute acceleration
        ay = -g  # Gravity acts downward
        
        # Euler update
        y  = y + vy * dt   # update height using current velocity
        vy = vy + ay * dt  # update velocity using acceleration
        t  = t + dt
        
        t_arr.append(t)
        y_arr.append(y)
        vy_arr.append(vy)
    
    return t_arr, y_arr, vy_arr
```

```python
h = 100
dt = 0.1

# Simulate the motion using the Euler method
# and compare it to the exact solution

t_arr, y_arr, vy_arr = simulate_freefall(h, dt)

t_exact = np.linspace(0, np.sqrt(2 * h / 9.81), 500)
y_exact = h - 0.5 * 9.81 * t_exact**2

plt.figure(figsize=(8, 5))
plt.plot(t_arr, y_arr, 'o-', label=f'Euler method (dt = {dt} s)')
plt.plot(t_exact, y_exact, 'k-', linewidth=2, label='Exact solution')
plt.xlabel('Time (s)')
plt.ylabel('Height y (m)')
plt.title('Free fall from height 100 m')
plt.grid(True)
plt.legend()
plt.show()
```
